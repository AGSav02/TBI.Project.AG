```python
import csv
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

shuffled_z = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_background_z1.tsv')
original_z = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_z1.tsv')

zscores_backgorund = []

with open(shuffled_z, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        zscores_backgorund.append(z_values)

zscores_data = []

with open(original_z, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        zscores_data.append(z_values)

zscores_backgorund = np.array(zscores_backgorund, dtype=float)
zscores_data = np.array(zscores_data, dtype=float)

kde_back = gaussian_kde(zscores_backgorund)
kde_org = gaussian_kde(zscores_data)

xmin = min(zscores_backgorund.min(), zscores_data.min())
xmax = max(zscores_backgorund.max(), zscores_data.max())
xgrid = np.linspace(xmin, xmax, 500)

plt.figure(figsize=(8,5))

density_back = kde_back(xgrid)
density_org = kde_org(xgrid)

plt.fill_between(xgrid, density_back, alpha=0.3, color="skyblue", label="Shuffled Z-scores")
plt.fill_between(xgrid, density_org, alpha=0.3, color="sandybrown", label="Original Z-scores")

plt.plot(xgrid, density_back, color="skyblue", linewidth=2)
plt.plot(xgrid, density_org, color="sandybrown", linewidth=2)


plt.xlabel("Z-score")
plt.ylabel("Density")
plt.title("Ditsribution of the Top1000 Ribosome Stalling Sites in Z1")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```
