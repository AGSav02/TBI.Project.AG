```python
import csv


output_tsv = "/scr/velara/bioinf2/TBI.Project.AG/master_folder_background_z1.tsv" 

zscore_list = []
header_saved = None

for n in range(1, 1001):
    files = f"/scr/velara/bioinf2/TBI.Project.AG/shuffling_command/z1_shuffled_data/{n}_shufled_z1.tsv"

    with open(files, 'r', newline='') as f:
        reader = csv.reader(f, delimiter='\t')

        header = next(reader, None)
        if header_saved is None:
            header_saved = header
        
        next(reader, None)

        for row in reader:
            zscore_list.append(row)
            
if zscore_list:
    with open(output_tsv, 'w', newline='') as f:
        writer = csv.writer(f, delimiter='\t')
        writer.writerow(header_saved)
        writer.writerows(zscore_list)
```

```python
import csv

input_files = [
    "/scr/velara/bioinf2/TBI.Project.AG/master_folder_background_z1.tsv",
    "/scr/velara/bioinf2/TBI.Project.AG/master_folder_background_z3.tsv",
    "/scr/velara/bioinf2/TBI.Project.AG/master_folder_background_z5.tsv"
]

output_file = "combined_background_mono.tsv"

all_rows = []
header_saved = None

for file_path in input_files:
    with open(file_path, "r", newline='') as f:
        reader = csv.reader(f, delimiter="\t")

        header = next(reader, None)
        if header_saved is None:
            header_saved = header 

        for row in reader:
            if row:  
                all_rows.append(row)

with open(output_file, "w", newline='') as f:
    writer = csv.writer(f, delimiter="\t")
    writer.writerow(header_saved)
    writer.writerows(all_rows)

```
