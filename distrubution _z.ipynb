```python
import csv
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

shuffled_z = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_background_z1.tsv')
original_z_di = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_z1.tsv')
original_z_mono = ('/scr/velara/bioinf2/TBI.Project.AG/master_folder_z1.tsv')

zscores_backgorund = []

with open(shuffled_z, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        zscores_backgorund.append(z_values)

zscores_data_di = []

with open(original_z_di, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        zscores_data_di.append(z_values)


zscores_data_mono = []

with open(original_z_mono, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        zscores_data_mono.append(z_values)

zscores_backgorund = np.array(zscores_backgorund, dtype=float)
zscores_data_di = np.array(zscores_data_di, dtype=float)
zscores_data_mono = np.array(zscores_data_mono, dtype=float)

kde_back = gaussian_kde(zscores_backgorund)
kde_org_di = gaussian_kde(zscores_data_di)
kde_org_mono = gaussian_kde(zscores_data_mono)

xmin = min(zscores_backgorund.min(), zscores_data_di.min(), zscores_data_mono.min())
xmax = max(zscores_backgorund.max(), zscores_data_di.max(), zscores_data_mono.max())
xgrid = np.linspace(xmin, xmax, 500)

plt.figure(figsize=(8,5))

density_back = kde_back(xgrid)
density_org_di = kde_org_di(xgrid)
density_org_mono = kde_org_mono(xgrid)

plt.fill_between(xgrid, density_back, alpha=0.3, color="skyblue", label="Shuffled Sequence Z-scores")
plt.fill_between(xgrid, density_org_di, alpha=0.3, color="sandybrown", label="Mono Shuffle Z-scores")
plt.fill_between(xgrid, density_org_mono, alpha=0.3, color="plum", label="Di Shuffle Z-scores")

plt.plot(xgrid, density_back, color="skyblue", linewidth=2)
plt.plot(xgrid, density_org_di, color="sandybrown", linewidth=2)
plt.plot(xgrid, density_org_mono, color="plum", linewidth=2)


plt.xlabel("Z-score")
plt.ylabel("Density")
plt.title("Z-Score Distribution (Z1): Background vs. Mono- and Dinucleotide Shuffles")
plt.legend()
plt.tight_layout()
plt.grid()
plt.show()
```
