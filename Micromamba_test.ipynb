```python
import RNA  #exmple in the vrna doc

sequence   = "CGGCGAAACGCAACGAAACGCG"
constraint = "((..................))"

fc = RNA.fold_compound(sequence)
fc.hc_add_from_db(constraint)
default_prediction = fc.mfe()

fc = RNA.fold_compound(sequence)
fc.hc_add_from_db(constraint, RNA.CONSTRAINT_DB_DEFAULT | RNA.CONSTRAINT_DB_ENFORCE_BP)
enforced_prediction = fc.mfe()

print(sequence)
print(constraint, "default")
print(f"{default_prediction[0]} {default_prediction[1]:.2f}")
print()

print(sequence)
print(constraint, "enforced")
print(f"{enforced_prediction[0]} {enforced_prediction[1]:.2f}")
print()
```

```python

with open('Homo_sapiens_ch7_GRCh38.fasta', 'r') as f:
    lines = f.readlines()

sequence = ""
for line in lines:
    if not line.startswith(">"):
        sequence += line.strip()

fc = RNA.fold_compound(sequence)
default_prediction = fc.mfe()

print(sequence)
print(f"{default_prediction[0]} {default_prediction[1]}")


```

```python
import varnaapi

v = varnaapi.Structure(sequence=sequence, structure=default_prediction[0])
v.show()
```

```python
import subprocess

input_fasta = "Homo_sapiens_ch7_GRCh38.fasta"
output_fasta = "Shuffled_ch7.fasta"

subprocess.run([
    "fasta-dinucleotide-shuffle-py3",
    "-f", input_fasta,
    "-s", "1",
    "-c", "300"
], stdout=open(output_fasta, "w"), check=True)

print(f"Shuffled sequences saved to {output_fasta}")

```
