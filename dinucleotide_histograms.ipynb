```python
import csv
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

data = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_z1.tsv')

zscores = []

with open(data, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        zscores.append(z_values)
    
    mean = np.mean(zscores)

    array_convert = np.array(zscores)
    kde = stats.gaussian_kde(array_convert)
    x_range = np.linspace(min(array_convert), max(array_convert), 200)

plt.figure(figsize=(10,4))
plt.hist(zscores, bins=40, edgecolor = 'black', color = 'skyblue', density='True')

plt.title("Z-score distribution of the Top 1000 Most Common Ribosome Stalling Site Sequences in Z1")
plt.xlabel('Z-scores')
plt.ylabel('Frequency of Z-scores Across Sequences')

plt.text(
    0.2, 0.98,
    f"Mean = {mean}",
    ha = 'left', va = 'top',
    transform=plt.gca().transAxes,
    fontsize=12,
    bbox=dict(facecolor='white', alpha=0.6)
)

plt.axvline(mean, color = "black", linestyle= 'dashed', linewidth = 1)
kde_y = kde(x_range)
plt.fill_between(x_range, kde_y, color='sandybrown', alpha=0.6)
plt.tight_layout()
plt.show()
```
