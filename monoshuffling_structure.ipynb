```python
import csv
import RNA

input_tsv = "/scr/velara/bioinf2/TBI.Project.AG/zscores_z1_out.tsv"
output_tsv = "z1_structure.tsv"

with open(input_tsv, "r") as infile, open(output_tsv, "w", newline="") as outfile:
    reader = csv.reader(infile, delimiter="\t")
    writer = csv.writer(outfile, delimiter="\t")

    header = next(reader)

    new_header = [header[1], header[2], "structure", header[0]]
    writer.writerow(new_header)

    for line_number, row in enumerate(reader, start=2):   
        if not row:
            continue

        sequence = row[0]
        mfe_tsv = float(row[1])      
        zscore = row[2]

        
        fc = RNA.fold_compound(sequence)
        structure, mfe_calc = fc.mfe()

        if mfe_calc != mfe_tsv:
            print(
                f"[ERROR] MFE mismatch at line {line_number}:\n"
                f"  Sequence: {sequence}\n"
                f"  TSV MFE: {mfe_tsv}\n"
                f"  Calculated MFE: {mfe_calc}\n"
            )

        new_row = [row[1], zscore, structure, sequence]
        writer.writerow(new_row)




```
