```python
import csv

output_tsv = "/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_z1.tsv" 

zscore_list = []
for n in range(1, 1001):
    files = f"/scr/velara/bioinf2/TBI.Project.AG/shuffling_command/z1_dishuffled_data/{n}_dishufled_z1.tsv"

    with open(files, 'r', newline='') as f:
        reader = csv.reader(f, delimiter='\t')
        header = next(reader)
        first_line = next(reader, None)

        if first_line:
            zscore_list.append(first_line)
            
if zscore_list:
    with open(output_tsv, 'w', newline='') as f:
        writer = csv.writer(f, delimiter='\t')
        writer.writerow(header)
        writer.writerows(zscore_list)
```
