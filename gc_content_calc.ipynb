```python
import csv

input_tsv = "/scr/velara/bioinf2/TBI.Project.AG/masterfolder_monoshuffling/master_folder_z1.tsv"
output_tsv = "z1_mono_gc.tsv"

def gc_content(seq):
    """Return GC content as a percentage."""
    seq = seq.upper()
    gc_count = seq.count("G") + seq.count("C")
    return round(gc_count / len(seq) * 100, 2) if len(seq) > 0 else 0

with open(input_tsv, "r") as infile, open(output_tsv, "w", newline="") as outfile:
    reader = csv.reader(infile, delimiter="\t")
    writer = csv.writer(outfile, delimiter="\t")

    header = next(reader)
    
    new_header = header[1:] + ["GC_content"] + [header[0]]
    writer.writerow(new_header)

    for row in reader:
        if not row:
            continue

        sequence = row[0]
        other_cols = row[1:]

        gc = gc_content(sequence)

        new_row = other_cols + [gc] + [sequence]
        writer.writerow(new_row)


```
