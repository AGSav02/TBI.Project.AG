```python
import matplotlib.pyplot as plt
import csv

data = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_z1.tsv')

seq_n = []
zscores = []

with open(data, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)

        seq_n.append(index)
        zscores.append(z_values)

plt.figure(figsize=(10,6))
bars = plt.bar(seq_n, zscores, color= ['skyblue'])
plt.xticks(rotation=45, ha='right')
plt.ylabel("Z-score Magnitude")
plt.xlabel("Sequence number in Top 1000 Most Common Sequences")
plt.title("Z-scores from the Top 1000 Most Common Ribosome Stalling Site Sequences in Z1")
plt.tight_layout()
plt.show()
```

```python
data = ('/scr/velara/bioinf2/TBI.Project.AG/dinucleotide_masterfolder_z1.tsv')

filtered_seq_n = []
filtered_zscores = []

with open(data, 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for index, row in enumerate(reader):
        if not row:
            continue

        raw_values = (row[2]).strip()
        z_values = float(raw_values)
        
        if z_values >2 or z_values < -2:
            filtered_seq_n.append(index)
            filtered_zscores.append(z_values)
        
        count = len(filtered_zscores)

plt.figure(figsize=(10, 4))
plt.bar(filtered_seq_n, filtered_zscores, color= 'skyblue')
plt.title("Z-scores with Values above 2 or below -2 (Outlier Comapred to 95.5% of the Data) in Z1")
plt.xlabel("Sequence number in Top 1000 Most Common Sequences")
plt.ylabel("Filtered Zscore Magnitude")
plt.text(
    0.98, 0.2,
    f"N = {count}",
    ha = 'right', va = 'top',
    transform=plt.gca().transAxes,
    fontsize=12,
    bbox=dict(facecolor='white', alpha=0.6)
)
plt.tight_layout()
plt.show()
```
