```python
import pandas as pd
import os

path = "/scr/velara/bioinf2/data/mrna_peaks/"
folders = ["z1", "z3", "z5"]

for folder_name in folders:
    folder_path = os.path.join(path, folder_name)
    output_tsv = f"{folder_name}_results.tsv"

    results = []
    transcript_dict = {}

    for file_name in os.listdir(folder_path):
        if not file_name.endswith(".bed3"):
            continue

        file_path = os.path.join(folder_path, file_name)
        if not os.path.isfile(file_path):
            continue

        with open(file_path, "r") as f:
            for line in f:
                parts = line.strip().split("\t")
                if len(parts) < 3:
                    continue

                transcript = parts[0]
                start = int(parts[1])
                end = int(parts[2])
                peak_range = (start, end)

                transcript_dict.setdefault(transcript, {})
                transcript_dict[transcript].setdefault(peak_range, set())

                transcript_dict[transcript][peak_range].add(file_name)

    for transcript, peaks in transcript_dict.items():
        most_common_peaks = [
            (peak_range, files)
            for peak_range, files in peaks.items()
            if len(files) >= 5
        ]

        if most_common_peaks:
            for peak_range, files in most_common_peaks:
                start, end = peak_range
                count = len(files)
                file_list = ",".join(sorted(files))
                results.append([transcript, file_list, f"{start}-{end}", int(count)])
        else:
            results.append([transcript, "N/A", "N/A", 0])
 
    df = pd.DataFrame(results, columns=["transcript_id", "file_name", "peak_position", "count"])
    df["count"] = pd.to_numeric(df["count"])
    df.sort_values(by = "count", ascending = False, inplace = True)
    df.to_csv(output_tsv, sep="\t", index=False)
    print(df.head()) 
```

```python
import pandas as pd
import glob
import os

tsv_path = glob.glob("scr/velara/bioinf2/TBI.Project.AG/results/*_1000.tsv")
bed_path = "/scr/velara/bioinf2/TBI.Project.AG/results/results_bedtools"

os.makedirs(bed_path, exist_ok=True)

for file in tsv_path:
    df = pd.read_csv(file, sep = "\t")
    df[['start', 'end']] = df['range'].str.split('-', expand = True)
    df['start'] = df['start'].astype(int)
    df['end'] = df['end'].astype(int)
    df['start'] = df['start'] - 1

    bed_df = df[['transcript_id', 'start', 'end']]

    base_name = os.path.splitext(os.path.basename(file))[0]
    output_path = os.path.join(output_dir, f'{base_name}.bed')

    bed_df.to_csv(output_path, sep = "\t", header = False, index = False)

    print(f"✅ Saved: {output_path}")
```
