```python
import numpy as np
import pandas as pd
import os

path = "/scr/velara/bioinf2/data/mrna_peaks/"
folders = ["z1", "z3", "z5"]

for name in folders:
    mrnapeaks_data = os.path.join(path, name)
    output_tsv = f"{name}_results.tsv"

    results = []
    transcript_dict = {}

    for file_name in os.listdir(mrnapeaks_data):
        if not file_name.endswith(".bed3"):
            continue
    
        file_path = os.path.join(mrnapeaks_data, file_name)
        if not os.path.isfile(file_path):
            continue
        
        with open(file_path, "r") as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) < 3:
                    continue
                transcript = parts[0]
                start = int(parts[1])
                end = int(parts[2])

                peak_range = (start, end)

                if transcript not in transcript_dict:
                    transcript_dict[transcript] = {}
            
                if peak_range not in transcript_dict[transcript]:
                    transcript_dict[transcript][peak_range] = 1
                else:
                    transcript_dict[transcript][peak_range] += 1

    for transcript, peaks in transcript_dict.items():
        most_common_peak = max(peaks.items(), key = lambda x: x[1])
        peak_range, count = most_common_peak
        if count > 1:
            start, end = peak_range
            results.append([transcript, f"{start}-{end}", count])
        else:
            results.append([transcript, "N/A", 0])

    df = pd.DataFrame(results, columns=["transcript_id", "peak_position", "count"])
    df.to_csv(output_tsv, sep="\t", index=False)
    print(df.head())
```
